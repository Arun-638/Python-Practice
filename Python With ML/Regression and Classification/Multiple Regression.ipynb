Multiple Regression using vectorization

```python
import numpy as np
x = np.array([[2104,5,1],[1416,3,2],[852,2,1]])
y = np.array([460, 232, 178])
w_init = np.zeros(3)
b_init = 0
```

```python
def compute_cost(x, y, w, b):
    m = x.shape[0]
    predictions = np.dot(x, w) + b
    errors = predictions - y
    cost = (1/(2*m)) * np.sum(errors ** 2)
    return cost
```

```python
def compute_gradient(x,y,w,b):
    m,n = x.shape
    predictions = np.dot(x,w)+b
    errors = predictions - y
    dj_dw = (1/m) * np.dot(x.T, errors) # here  x was m,n and errors was m, so we need to take the transpose of x to get n,m and then multiply with m,1 to get n,1
    dj_db = (1/m) * np.sum(errors)
    return dj_dw, dj_db
```

```python
def gradient_descent(x,y,w_in,b_in,alpha,iterations):
    w = w_in
    b = b_in
    for i in range(iterations):
        dj_dw,dj_db = compute_gradient(x,y,w,b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
    return w, b
```

```python
alpha = 5.0e-6
iterations = 100
print("starting w:", w_init, "b:", b_init, "cost:", compute_cost(x, y, w_init, b_init))
print("Running gradient descent...")
w_final, b_final = gradient_descent(x, y, w_init, b_init, alpha, iterations)
print("After running gradient descent w:", w_final, "b:", b_final, "cost:", compute_cost(x, y, w_final, b_final))
```

Here we can see the cost is approx 2 by reducing the learning rate drastically... so this problem can be overcomed by feature scaling
